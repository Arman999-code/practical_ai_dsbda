```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Set seed
np.random.seed(10)

# Create dataset
data = {
    'Student_ID': range(1, 51),
    'Gender': np.random.choice(['Male', 'Female'], 50),
    'Math_Score': np.random.normal(70, 15, 50),
    'Science_Score': np.random.normal(65, 12, 50),
    'English_Score': np.random.normal(75, 10, 50),
    'Attendance_Percentage': np.random.normal(85, 8, 50),
    'Study_Hours_per_Week': np.random.normal(15, 5, 50)
}

df = pd.DataFrame(data)

# Introduce missing values and inconsistencies
df.loc[5, 'Math_Score'] = np.nan
df.loc[10, 'Science_Score'] = np.nan
df.loc[15, 'Attendance_Percentage'] = np.nan
df.loc[20, 'Gender'] = 'male'   # inconsistent case
df.loc[25, 'Math_Score'] = 200  # outlier

# Check missing values
print("Missing values before cleaning:")
print(df.isnull().sum())

# Fill missing values
df['Math_Score'].fillna(df['Math_Score'].mean(), inplace=True)
df['Science_Score'].fillna(df['Science_Score'].mean(), inplace=True)
df['Attendance_Percentage'].fillna(df['Attendance_Percentage'].mean(), inplace=True)

# Fix categorical inconsistency
df['Gender'] = df['Gender'].str.capitalize()

# Handle outliers using IQR method
Q1 = df['Math_Score'].quantile(0.25)
Q3 = df['Math_Score'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df['Math_Score'] = np.where(
    df['Math_Score'] < lower_bound,
    lower_bound,
    np.where(df['Math_Score'] > upper_bound, upper_bound, df['Math_Score'])
)

# Skewness before transformation
print("\nSkewness before log transform:")
print(df['Study_Hours_per_Week'].skew())

# Log transformation
df['Log_Study_Hours'] = np.log(df['Study_Hours_per_Week'] + 1)

# Output results
print("\nFirst 10 rows of cleaned dataset:")
print(df.head(10))

print("\nDataset info:")
print(df.info())
```
